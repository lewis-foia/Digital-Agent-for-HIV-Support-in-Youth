### Knowledge Base

````python
#The hiv_faq.txt was provided in the project folder

kb_path = "/kaggle/working/hiv_faq.txt"

with open(kb_path, "r") as f:
    text_data = f.read().split("\n\n")
print(f"Loaded {len(text_data)} knowledge chunks.")
````

````python
# FAISS Index Building
from sentence_transformers import SentenceTransformer
import faiss, numpy as np

embedder = SentenceTransformer("all-MiniLM-L6-v2")

# Text conversion into embbedings
embs = embedder.encode(text_data)
index = faiss.IndexFlatL2(embs.shape[1])
index.add(np.array(embs))

def retrieve(query, k=2):
    q_emb = embedder.encode([query])
    D, I = index.search(np.array(q_emb), k)
    return [text_data[i] for i in I[0]]
````

### Loading the LLM

````python
#Cohere Command A 03 2025
import cohere

API_KEY = "GET_YOUR_API_KEY_AND_ADD_IT_HERE"

co = cohere.Client(API_KEY)

co = cohere.Client(API_KEY)
print("✅ Cohere client ready")

def ask_agent(query):
    docs = retrieve(query, k=2)
    context = " ".join(docs)
    response = co.chat(
        model="command-a-03-2025",
        message=f"Context: {context}\nUser: {query}\nAnswer in a clear and supportive way:",
        temperature=0.3,
        p=0.9
    )
    return response.text
````

### Example Test + Appointment Scheduler

````python
print(ask_agent("What is ART treatment?"))
print(ask_agent("Can I get HIV from hugging?"))
````

````python
def schedule_appointment(date):
    return f"✅ Appointment scheduled for {date}. Please remember to attend and bring your treatment records."
````

### Chat Loop

````python
def chat_loop():
    print("🤖 HIV Support Agent (type 'quit' to stop)\n")
    while True:
        user = input("You: ")
        if user.lower() in ["quit", "exit", "stop"]:
            break
        if "appointment" in user.lower():
            if "on" in user.lower():
                    date = user.split("on")[-1].strip()
                    print("Bot:", schedule_appointment(date))
            else:
                    print("Bot: Please specify a date for the appointment.")
        else:
            print("Bot:", ask_agent(user))
        print()

chat_loop()
````
